```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
data = pd.read_csv("ai_model_experiments_5000.csv")
data.info()
```

```python
plt.figure()
heatmap_data = data.drop(columns="experiment_id")
corr = heatmap_data.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
```

```python
plt.figure()
sns.pairplot(data, hue="Model")
```
